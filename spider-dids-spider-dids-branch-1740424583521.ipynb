## Multiple Sequence Alignment using MAFFT

This Jupyter Notebook performs Multiple Sequence Alignment (MSA) using MAFFT on selected sequences and visualizes the results using Plotly. You can also open the alignment in Jalview.

```python
# Import necessary libraries
import os
import numpy as np
import pandas as pd
import plotly.express as px
from Bio import SeqIO
```

### Prepare FASTA File for Alignment

```python
# Define the sequence file
fasta_file = 'sequences.fasta'
with open(fasta_file, 'w') as f:
    sequences = [
        ">Agelenidae_Agelena_silvatica\nATGAAGTTTATAACTGATTTGCAATCAGT-TAATATGTTGT----ACATACTTCAAT\n",
        ">Araneidae_Araneus_angulatus\nTATAAAGTTTTAACTGATTTGCAATCAGG-ATATATGAAAT----TCATACTAGAAT\n",
        ">Araneidae_Araneus_ventricosus\nTTGAGAGTTTAAGTTGATTTGCAATCAGC-TAGAATGTTTA----ACATACTTTGAA\n"    ]
    f.writelines(sequences)
print(f'Sequences saved to {fasta_file}')
```

### Run MAFFT Alignment

```python
# Run MAFFT for Multiple Sequence Alignment
aligned_file = 'aligned_sequences.fasta'
!mafft --auto sequences.fasta > aligned_sequences.fasta
print(f'Alignment completed! Results saved to {aligned_file}')
```

### View Aligned Sequences

```python
# Read and display aligned sequences
with open(aligned_file, 'r') as f:
    print(f.read())
```

### Visualize MSA with Plotly

```python
# Function to convert sequences into a DataFrame for visualization
def sequences_to_dataframe(aligned_file):
    sequences = list(SeqIO.parse(aligned_file, 'fasta'))
    df = pd.DataFrame(
        [list(str(seq.seq)) for seq in sequences],
        index=[seq.id for seq in sequences]
    )
    return df

# Convert MSA to DataFrame
msa_df = sequences_to_dataframe(aligned_file)

# Convert characters to numeric values for heatmap
unique_chars = list(set(msa_df.values.flatten()))
char_to_num = {char: i for i, char in enumerate(unique_chars)}
msa_numeric = msa_df.replace(char_to_num)

# Plot heatmap
fig = px.imshow(msa_numeric, color_continuous_scale='viridis', labels={'x': 'Position', 'y': 'Sequence'})
fig.update_layout(title='Multiple Sequence Alignment (MSA) Heatmap', width=900, height=500)
fig.show()
```
